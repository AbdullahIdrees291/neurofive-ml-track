**The Titanic dataset has 891 rows and 12 columns. Some columns, such as Age, Cabin, and Embarked, contain missing values. Numerical columns include Age, Fare, and PassengerId, while categorical columns include Name, Sex, Ticket, Cabin, and Embarked. The dataset needs cleaning before it can be used for machine learning.**

```python
df['Age'].fillna(df['Age'].median(), inplace=True)

df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

I handled missing values by filling Age with the median because age is numerical and the median is less affected by outliers. I filled Embarked with the most common value because it is a categorical feature.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.boxplot(x=df['Age'])
plt.title("Age Outliers")
plt.show()
```

```python
sns.histplot(df['Age'], bins=30)
plt.title("Age Distribution")
plt.show()
```

```python
sns.boxplot(x=df['Fare'])
plt.title("Fare Outliers")
plt.show()
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

The feature that most affects survival is Sex. From the analysis, females had a higher survival rate compared to males. This suggests that gender played an important role in survival during the Titanic disaster.
